```python
corr_columns = [
    "burnout_score",
    "stress_score",
    "work_life_balance_score",
    "manager_support_score",
    "job_satisfaction_score",
    "deadline_pressure_score",
    "autonomy_score",
    "sleep_hours_per_night",
    "exercise_days_per_week",
    "work_hours_per_week",
    "meetings_per_day",
    "years_at_company",
    "salary_usd",
    "phq9_score",
    "gad7_score"
]

burnout_corr = (
    df[corr_columns]
    .corr()["burnout_score"]
    .sort_values(ascending=False)
)

print(burnout_corr)
```

```python

driver_ranking = burnout_corr.drop("burnout_score")


driver_ranking = driver_ranking.reset_index()
driver_ranking.columns = ["Factor", "Correlation"]


driver_ranking["Absolute_Correlation"] = driver_ranking["Correlation"].abs()


driver_ranking = driver_ranking.sort_values(
    by="Absolute_Correlation",
    ascending=False
)

driver_ranking
```

```python
def relationship(x):
    if x > 0:
        return "Positive Association"
    else:
        return "Negative Association"

driver_ranking["Association"] = driver_ranking["Correlation"].apply(relationship)

driver_ranking
```

```python
df["sleep_category"] = pd.cut(
    df["sleep_hours_per_night"],
    bins=[0,5,6,7,10],
    labels=[
        "Less than 5 hrs",
        "5-6 hrs",
        "6-7 hrs",
        "More than 7 hrs"
    ]
)

sleep_analysis = (
    df.groupby("sleep_category", observed=False)["burnout_score"]
      .mean()
      .reset_index()
)
sleep_analysis
```

```python
df["manager_support_level"] = pd.cut(
    df["manager_support_score"],
    bins=[0,4,7,10],
    labels=[
        "Low Support",
        "Medium Support",
        "High Support"
    ]
)

manager_analysis = (
    df.groupby("manager_support_level",observed=False)["burnout_score"]
      .mean()
      .reset_index()
)

manager_analysis
```

```python
import pandas as pd
import numpy as np

# ---------- PERSONA ----------
df["persona"] = "Other"

# Flight Risk
df.loc[
    (df["burnout_level"].isin(["High", "Severe"])) &
    (df["job_change_intention"] == 1),
    "persona"
] = "Flight Risk"

# Early Risk
df.loc[
    (df["persona"] == "Other") &
    (df["years_at_company"] < 2) &
    (df["burnout_level"].isin(["High", "Severe"])),
    "persona"
] = "Early Risk"

# Silent Burnout
df.loc[
    (df["persona"] == "Other") &
    (df["burnout_level"].isin(["High", "Severe"])),
    "persona"
] = "Silent Burnout"

# Healthy Performer
df.loc[
    (df["persona"] == "Other") &
    (df["burnout_level"] == "Low") &
    (df["job_satisfaction_score"] >= 7),
    "persona"
] = "Healthy Performer"

# Resilient
df.loc[
    (df["persona"] == "Other") &
    (df["years_at_company"] >= 5) &
    (df["burnout_level"] == "Low"),
    "persona"
] = "Resilient"

# Sleep 
df["sleep_category"] = pd.cut(
    df["sleep_hours_per_night"],
    bins=[0, 5, 6, 7, 10],
    labels=[
        "Less than 5 hrs",
        "5-6 hrs",
        "6-7 hrs",
        "More than 7 hrs"
    ]
)

# manager support
df["manager_support_level"] = pd.cut(
    df["manager_support_score"],
    bins=[0, 4, 7, 10],
    labels=[
        "Low Support",
        "Medium Support",
        "High Support"
    ]
)

print(" New columns created successfully!")
```
